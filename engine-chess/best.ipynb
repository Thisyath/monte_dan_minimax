```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Data
metrics = ["Avg Move Time (s)", "Avg CP Loss", "Best Move Match (%)"]
hybrid_low = [2.0000, 567.57, 30.50]
hybrid_high = [2.0000, 381.51, 38.19]

# Buat DataFrame
df = pd.DataFrame({
    "Metric": metrics,
    "Hybrid Low": hybrid_low,
    "Hybrid High": hybrid_high
})

# Fungsi untuk menambahkan label di atas bar
def add_labels(x, y):
    for i in range(len(x)):
        plt.text(i, y[i] + (max(y)*0.02), round(y[i],2), ha='center', fontsize=10)

# 1. Performance comparison: Move Time
plt.figure(figsize=(6,4))
values = [hybrid_low[0], hybrid_high[0]]
bars = plt.bar(["Hybrid Low","Hybrid High"], values, color=["skyblue","steelblue"])
plt.title("Performance Comparison: Avg Move Time")
plt.ylabel("Seconds")
add_labels(["Hybrid Low","Hybrid High"], values)
plt.show()

# 2. Accuracy (lower is better): Avg CP Loss
plt.figure(figsize=(6,4))
values = [hybrid_low[1], hybrid_high[1]]
bars = plt.bar(["Hybrid Low","Hybrid High"], values, color=["lightgreen","green"])
plt.title("Accuracy Comparison: Avg CP Loss (lower is better)")
plt.ylabel("Avg CP Loss")
add_labels(["Hybrid Low","Hybrid High"], values)
plt.show()

# 3. Move Matching (%)
plt.figure(figsize=(6,4))
values = [hybrid_low[2], hybrid_high[2]]
bars = plt.bar(["Hybrid Low","Hybrid High"], values, color=["orange","darkorange"])
plt.title("Best Move Match (%)")
plt.ylabel("Percentage")
add_labels(["Hybrid Low","Hybrid High"], values)
plt.show()

# Ringkasan Data
print("Ringkasan Data:")
print(df)

```
